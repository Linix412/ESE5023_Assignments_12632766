```python
#1.Flowchart
import random
def Print_values():
    a = round(random.random()*10)
    b = round(random.random()*10)
    c = round(random.random()*10)
#保证a、b、c三个数不会相等
    while b == a:
        b = round(random.random()*10)
    while c == b or c==a:
        c = round(random.random()*10)
        
    if a > b:
        if b > c:
            print("a =",a,">","b =",b,">","c =",c)
        else:
            if a > c:
                print("a =",a,">","c =",c,">","b =",b)
            else:
                print("c =",c,">","a =",a,">","b =",b)
    else:
        if b > c:
            print("a =",a,"b =",b,"c =",c)
            print("done")
        else:
            print("c =",c,">","b =",b,">","a =",a)
Print_values()
```

```python
#2. Matrix multiplication
#2.1
import random 
import numpy as np 
M1 = np.random.randint(50,size=(5,10)) 
M2 = np.random.randint(50,size=(10,5)) 
print(M1) 
print(M2)
#2.2
def Matrix_multip():
    x = 0
    i = 0
    j = 0
    z = 0
    M3 = np.array([])
    for i in range(0,5):
        for j in range(0,5):
            for x in range(0,10):
                z = z + M1[i,x] * M2[x,j]
                x = x+1
            M3 = np.append(M3,z)
            z = 0
            j = j+1
        i = i+1
    print(M3.reshape(5,5))
Matrix_multip()
                
                
```

```python
#3. Pascal triangle
def Pascal_triangle():
    k = int(input("tell me the figure(>=2) :"))#询问输出第几行
    l = [1]
    #print(l)#展示第0行
    l.append(1)
    #print(l)#展示第1行
    n = l.copy()#暂时作为新行
    x = 1 #用于计数，要为下一行行计算多少个新元素
    i = 1
    while x <= k-1:
        for i in range(1,x+1):
            n.insert(int(i),l[i-1]+l[i])
            i = i+1
        l = n.copy()
        n.clear()
        n = [1,1]
        #print(l)#展示当前行
        x = x+1
    print(l)#展示第k行
Pascal_triangle()
#本代码结果最少从杨辉三角的第一行开始展示，若输入k小于等于0，输出结果也只为第一行
```

```python
#4. Add or double
import random
def Least_moves():
    x = random.randint(1, 100)
    print("x=" + str(x))
    y = 0
    while x > 1:
        if x % 2 == 0:#识别x是否为偶数
            x = x/2
        else:
            x = x-1
        y = y+1#记录移动步数
    print("at least move "+str(y)+" times")
        
Least_moves()
```

```python
#5. Dynamic programming
#5.1

#只能随机生成一个式子，后续可通过在外层加一个循环来得到一个或多个正确的结果，但会漏解

import random
def Find_expression(x):
    evaluate1 = ["+","-",""]
    evaluate2 = ["+","-",]
    i = 0
    l = []
    for i in range(8):
        if l and l[-1] == "":
            c = random.choice(evaluate2)
        else:
            c = random.choice(evaluate1)
        l.append(c)
    #print(l)
    
    digits = "123456789"
    expression = digits[0]
    i = 0
    for y in l:
        expression = expression + y + digits[i+1]
        i = i+1
    print(expression)
    
    y = eval(expression) #用eval计算等式结果
    print(expression,"=",y)
 
Find_expression(x)
```

```python
#5. Dynamic programming
#5.1
from itertools import product
def Find_expression(x,show = False):
    digits = "123456789"
    c = 0
    for combo in product(['+', '-', ''], repeat=8):
        expression = digits[0]
        for i, calculation in enumerate(combo):
            expression = expression + calculation + digits[i+1]
        
        value = eval(expression)#计算等式结果
        if value == x:
            if show:#用于判断是否展示算式
                print(expression + "=" + str(x))  #展示符合要求的算式
            c = c+1
    return c

Find_expression(50,show=True)

#5.2
l = []
for i in range(1, 101):
    count = Find_expression(i)
    l.append(count)
print("最大值：", max(l),"对应数字：", l.index(max(l)) + 1)
print("最小值：", min(l),"对应数字：", l.index(min(l)) + 1)
```
